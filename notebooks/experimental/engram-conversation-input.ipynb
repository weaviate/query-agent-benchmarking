# ConversationInput Demo

This notebook illustrates how to use Engram's `ConversationInput` when adding memories.

```python
from engram import EngramClient

client = EngramClient(
    api_key=os.getenv("ENGRAM_API_KEY"),
    base_url="https://dev-engram.labs.weaviate.io"
)
```

```python
from engram import ConversationInput, MessageInput

run = client.memories.add(
    ConversationInput(
        messages=[
            MessageInput(role="user", content="What's the best way to handle retries?"),
            MessageInput(role="assistant", content="Exponential backoff with jitter is the standard approach."),
            MessageInput(role="user", content="Got it — I'll use that in my HTTP client."),
        ]
    ),
    user_id="user_321",
)
```
