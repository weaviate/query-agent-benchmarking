```python
import os
```

```python
import weaviate
weaviate_client = weaviate.connect_to_weaviate_cloud(
    cluster_url=os.environ["WEAVIATE_URL"],
    auth_credentials=weaviate.AuthApiKey(api_key=os.environ["WEAVIATE_API_KEY"])
)
```

```python
weaviate_client.collections.list_all()["IRPapersImages_Default"]
```

```python
from weaviate.classes.config import Reconfigure

irpapers_images = weaviate_client.collections.get("IRPapersImages_Default")

irpapers_images.config.update(
    vector_config=Reconfigure.Vectors.update(
        name="default",  # name of the vectorizer / named vector
        vector_index_config=Reconfigure.VectorIndex.hnsw(
            ef=512,
        ),
    ),
)
```
