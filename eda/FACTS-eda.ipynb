```python
from datasets import load_dataset

ds = load_dataset("google/FACTS-grounding-public", "examples")
```

```python
ds
```

```python
print("System Instruction:")
print(ds["public"][0]["system_instruction"])
print("="*100)
print("User Request:")
print(ds["public"][0]["user_request"])
print("="*100)
print("Context Document:")
print(ds["public"][0]["context_document"])
```

```python
import tiktoken

# Choose an encoding, e.g., for OpenAI's GPT-3.5/4
enc = tiktoken.encoding_for_model("gpt-3.5-turbo")

token_counts = []
for example in ds["public"]:
    context = example["context_document"]
    token_counts.append(len(enc.encode(context)))

print(f"Total documents: {len(token_counts)}")
print(f"Average tokens per document: {sum(token_counts) / len(token_counts):.2f}")
print(f"Min tokens: {min(token_counts)}, Max tokens: {max(token_counts)}")
```
